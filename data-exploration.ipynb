```python
import pandas as pd
```

```python
books = pd.read_csv(f"{path}/books.csv")
books
```

```python
# checking is the missing values have a pattern which could indicate a bias
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)
plt.xlabel("Missing vals")
plt.ylabel("Columns")
```

categories, thumbnails and description is random
but the last three has a pattern there. So probably from a different dataset. And that dataset didnt include the books from the originial dataset. This will be explored next:

```python
import numpy as np

books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"]  = 2025 - books["published_year"]
```

```python
columns_of_interest = ["num_pages" , "age_of_book", "missing_description", "average_rating"]

# missing description is binary spearman is better when we are dealing with non-cont variables
corr_matrix = books[columns_of_interest].corr(method = "spearman")

sns.set_theme(style="whitegrid")
plt.figure(figsize=(6, 6))
heatmap = sns.heatmap(corr_matrix, annot=True, cmap="coolwarm",
                      cbar_kws={"label": "Spearman Correlation"})
heatmap
```

missing descriptions do not seem to have any correlations with anything. So in general we have representative descriptions.

```python
# checking the number of na observations i will remove. Only 300 rows so ok to remove.
books[(books["description"].isna())|
      (books["num_pages"].isna())|
      (books["average_rating"].isna())|
      (books["published_year"].isna())
        ]
```

```python
missing_books = books[(~books["description"].isna()) &
      (~books["num_pages"].isna()) &
      (~books["published_year"].isna()) &
      (~books["average_rating"].isna())]
```

```python
missing_books
```

```python
missing_books["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

with the bar chart visualization, i saw that there is imbalance in categories. Some categories have only have 1 or 2 books. Long tail problem. This will be solved with the use of LLMs.

```python
# some descriptions are 1 word so useless.
missing_books["words_in_description"] = missing_books["description"].str.split().str.len()
```

```python
missing_books.loc[missing_books["words_in_description"].between(25,34), "description"]
```

```python
book_missing_25_words = missing_books[missing_books["words_in_description"] >= 25]
book_missing_25_words
```

```python
# creating a new colum title_and_subtitle
book_missing_25_words["title_and_subtitle"] = (
    # when subtitle is missing
    np.where(book_missing_25_words["subtitle"].isna(), book_missing_25_words["title"],
             book_missing_25_words[["title", "subtitle"]].astype(str).agg(": ".join, axis=1))
)
book_missing_25_words
```

```python
#tags each description with a unique identifier. This will be useful for building the vector database.
book_missing_25_words["tagged_description"] = book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
```

```python
book_missing_25_words
```

```python
# drop columns that are not needed (only used for data cleaning and filtering)
(
    book_missing_25_words
    .drop(columns=["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
    .to_csv("data_cleaned.csv", index=False)
 )
```
